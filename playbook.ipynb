# Минимальный запрос к локальной модели через ZEMI Arsenal

Этот пример запускает самую маленькую модель из curated set в Model Mode, получает настроенный OpenAI-клиент через Arsenal и отправляет модели короткое сообщение.

## Подготовка компонента

```python
# Automatically reload imported modules when their source code changes
%load_ext autoreload
%autoreload 2

# Set the working directory to the ZEMI component root
from pathlib import Path

while not Path(".zemicomp").is_file():
    if Path.cwd().parent == Path.cwd():
        raise FileNotFoundError("Could not find the ZEMI component root")
    %cd ..
PROJECT_ROOT = Path.cwd()
PROJECT_ROOT
```

## Запуск Arsenal в Model Mode

```python
import zemi
from zemi.arsenal import ArsenalSession


arsenal = ArsenalSession(
    "@comp/zemi/llm_curated_set_model_mode.toml"
)
zemi.arsenal.begin(
    arsenal,
    stop_before_begin=True,
    llama_router_mode=False,
)
```

## Запрос к модели

```python
assistant = (
    arsenal.llamas["lfm2_350m_server"]
    .models["lfm2_350m"]
    .assistants["assistant"]
)
client = assistant.clients.openai.client
```

```python
response = client.chat.completions.create(
    model=assistant.clients.model,
    messages=[
        {
            "role": "user",
            "content": "Привет!",
        },
    ],
)
print(response.choices[0].message.content)
```

## Остановка Arsenal

После запроса завершаем сессию и останавливаем локальный сервер.

```python
zemi.arsenal.end(
    arsenal,
    stop_after_end=True,
)
```
